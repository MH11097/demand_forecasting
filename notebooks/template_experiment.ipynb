# Template thử nghiệm

Khung trống để thử nhanh 1 model trên Favorita. Đổi `MODEL` rồi chạy.

```python
import sys; sys.path.insert(0, '..')
from src.utils.config import load_config
from src.data.loader import load_raw_data, select_series
from src.data.features import add_all_features, apply_log_transform
from src.data.preprocessor import preprocess
from src.models import get_model_class
```

```python
MODEL = 'xgboost'
config = load_config(MODEL)
df, _ = load_raw_data(config)
df = select_series(df, config)
df = add_all_features(df, feature_cfg=config.get('features'))
if config.get('use_log_sales'): df = apply_log_transform(df)
train_df, val_df, test_df, _ = preprocess(df, config)
```

```python
model = get_model_class(MODEL)(config)
model.train(train_df, val_df)
model.evaluate(test_df)   # dict NWRMSLE/RMSE/MAE/MAPE
```
